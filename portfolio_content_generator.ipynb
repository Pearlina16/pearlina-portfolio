```python
# Export All Configuration Data
def generate_js_content():
    """Generate the complete JavaScript content for content_option.js"""
    
    js_content = f'''const logotext = "{logotext}";
const meta = {{
    title: "{meta['title']}",
    description: "{meta['description']}",
}};

const introdata = {{
    title: "{introdata['title']}",
    animated: {{
        first: "{introdata['animated']['first']}",
        second: "{introdata['animated']['second']}",
        third: "{introdata['animated']['third']}",
    }},
    description: "{introdata['description']}",
    your_img_url: "{introdata['your_img_url']}",
}};

const dataabout = {{
    title: "{dataabout['title']}",
    aboutme: `{dataabout['aboutme']}`,
}};

const worktimeline = {json.dumps(worktimeline, indent=4)};

const skills = {json.dumps(skills, indent=4)};

const services = {json.dumps(services, indent=4)};

const dataportfolio = {json.dumps(dataportfolio, indent=4)};

const contactConfig = {{
    YOUR_EMAIL: "{contactConfig['YOUR_EMAIL']}",
    YOUR_FONE: "{contactConfig['YOUR_FONE']}",
    description: "{contactConfig['description']}",
    YOUR_SERVICE_ID: "{contactConfig['YOUR_SERVICE_ID']}",
    YOUR_TEMPLATE_ID: "{contactConfig['YOUR_TEMPLATE_ID']}",
    YOUR_USER_ID: "{contactConfig['YOUR_USER_ID']}",
}};

const socialprofils = {{
    github: "{socialprofils['github']}",
    facebook: "{socialprofils['facebook']}",
    linkedin: "{socialprofils['linkedin']}",
    twitter: "{socialprofils['twitter']}",
}};

export {{
    meta,
    dataabout,
    dataportfolio,
    worktimeline,
    skills,
    services,
    introdata,
    contactConfig,
    socialprofils,
    logotext,
}};'''
    
    return js_content

# Generate the content
js_content = generate_js_content()

print("✅ JavaScript content generated successfully!")
print(f"📄 Content length: {len(js_content)} characters")
print("🎯 Ready to copy and paste into your content_option.js file!")
```
